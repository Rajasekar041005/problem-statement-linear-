```python
import pandas as pd
df = pd.read_csv('/content/cars_fuel_efficiency_clean.csv')
display(df.head())
```

```python
df.info()

display(df.describe())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np

X_mileage = df[['distance_km', 'avg_speed_kmph', 'trip_duration_hr']]
y_mileage = df['actual_mileage_kmpl']

X_mileage_train, X_mileage_test, y_mileage_train, y_mileage_test = train_test_split(X_mileage, y_mileage, test_size=0.2, random_state=42)

mileage_model = LinearRegression()
mileage_model.fit(X_mileage_train, y_mileage_train)

y_mileage_pred = mileage_model.predict(X_mileage_test)

mse_mileage = mean_squared_error(y_mileage_test, y_mileage_pred)
r2_mileage = r2_score(y_mileage_test, y_mileage_pred)

print(f"Mileage Prediction Model - Mean Squared Error: {mse_mileage:.2f}")
print(f"Mileage Prediction Model - R-squared: {r2_mileage:.2f}")
print(f"Mileage Prediction Model - Coefficients: {mileage_model.coef_}")
print(f"Mileage Prediction Model - Intercept: {mileage_model.intercept_:.2f}")
```

```python
X_fuel = df[['distance_km', 'avg_speed_kmph', 'base_mileage_kmpl']]
y_fuel = df['fuel_consumed_L']

X_fuel_train, X_fuel_test, y_fuel_train, y_fuel_test = train_test_split(X_fuel, y_fuel, test_size=0.2, random_state=42)

fuel_model = LinearRegression()
fuel_model.fit(X_fuel_train, y_fuel_train)

y_fuel_pred = fuel_model.predict(X_fuel_test)

mse_fuel = mean_squared_error(y_fuel_test, y_fuel_pred)
r2_fuel = r2_score(y_fuel_test, y_fuel_pred)

print(f"Fuel Consumption Prediction Model - Mean Squared Error: {mse_fuel:.2f}")
print(f"Fuel Consumption Prediction Model - R-squared: {r2_fuel:.2f}")
print(f"Fuel Consumption Prediction Model - Coefficients: {fuel_model.coef_}")
print(f"Fuel Consumption Prediction Model - Intercept: {fuel_model.intercept_:.2f}")
```

```python
def predict_actual_mileage(distance_km, avg_speed_kmph, trip_duration_hr):
    input_data = pd.DataFrame([[distance_km, avg_speed_kmph, trip_duration_hr]], columns=['distance_km', 'avg_speed_kmph', 'trip_duration_hr'])
    return mileage_model.predict(input_data)[0]

def predict_fuel_consumption(distance_km, avg_speed_kmph, base_mileage_kmpl):
    input_data = pd.DataFrame([[distance_km, avg_speed_kmph, base_mileage_kmpl]], columns=['distance_km', 'avg_speed_kmph', 'base_mileage_kmpl'])
    return fuel_model.predict(input_data)[0]

def calculate_remaining_distance(current_fuel_L, predicted_mileage_kmpl):
    if predicted_mileage_kmpl <= 0:
        return 0
    return current_fuel_L * predicted_mileage_kmpl

current_fuel = 30  # Liters
example_trip_distance = 150 # km
example_avg_speed = 70 # km/h
example_trip_duration = 2 # hours
example_base_mileage = 15 # km/l

predicted_mileage = predict_actual_mileage(example_trip_distance, example_avg_speed, example_trip_duration)
print(f"Predicted Actual Mileage for the trip: {predicted_mileage:.2f} km/l")

remaining_distance = calculate_remaining_distance(current_fuel, predicted_mileage)
print(f"Remaining distance vehicle can travel with {current_fuel}L fuel: {remaining_distance:.2f} km")
```

```python
def optimize_delivery_trip(trip_distance, avg_speed, trip_duration, base_mileage, current_fuel_L):
    predicted_mileage = predict_actual_mileage(trip_distance, avg_speed, trip_duration)

    predicted_fuel_consumed = predict_fuel_consumption(trip_distance, avg_speed, base_mileage)

    fuel_after_trip = current_fuel_L - predicted_fuel_consumed
    remaining_distance_after_trip = calculate_remaining_distance(fuel_after_trip, predicted_mileage)

    trip_feasible = current_fuel_L >= predicted_fuel_consumed

    return {
        "predicted_actual_mileage_kmpl": predicted_mileage,
        "predicted_fuel_consumed_L": predicted_fuel_consumed,
        "fuel_after_trip_L": fuel_after_trip,
        "remaining_distance_after_trip_km": remaining_distance_after_trip,
        "trip_feasible": trip_feasible
    }


print("\n--- Scenario 1: Short, Fast Trip ---")
scenario1_results = optimize_delivery_trip(trip_distance=50, avg_speed=80, trip_duration=0.6, base_mileage=15, current_fuel_L=20)
for key, value in scenario1_results.items():
    print(f"{key}: {value:.2f}" if isinstance(value, (float, np.float64)) else f"{key}: {value}")

print("\n--- Scenario 2: Longer, Moderate Speed Trip ---")
sceanrio2_results = optimize_delivery_trip(trip_distance=200, avg_speed=60, trip_duration=3.3, base_mileage=15, current_fuel_L=20)
for key, value in sceanrio2_results.items():
    print(f"{key}: {value:.2f}" if isinstance(value, (float, np.float64)) else f"{key}: {value}")

print("\n--- Scenario 3: Very Long Trip ---")
scenario3_results = optimize_delivery_trip(trip_distance=400, avg_speed=70, trip_duration=5.7, base_mileage=15, current_fuel_L=20)
for key, value in scenario3_results.items():
    print(f"{key}: {value:.2f}" if isinstance(value, (float, np.float64)) else f"{key}: {value}")
```
