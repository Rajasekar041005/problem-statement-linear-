Multiclass classification

```python
from sklearn.model_selection import train_test_split
import pandas as pd
df=pd.read_csv("/content/cars_fuel_efficiency_clean.csv")
df
```

```python
X = df.drop(['efficiency_band', 'trip_id', 'car_name'], axis=1)
y = df['efficiency_band']
```

```python
x=pd.get_dummies(X)
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
model=LogisticRegression()
model.fit(x_train,y_train)
```

```python
model.score(x_test,y_test)
```

```python
model.predict(x_test)
```

Decision Tree

```python
from sklearn.tree import DecisionTreeClassifier
model=DecisionTreeClassifier(random_state=42)
model.fit(x_train,y_train)
```

```python
model.score(x_test,y_test)
```

```python
model.predict(x_test)
```

Regularization

```python
model_C=SVC(C=1)
model_C.fit(x_train,y_train)
model_C.score(x_test,y_test)
```

Gamma

```python
model_g=SVC(gamma=10)
model_g.fit(x_train,y_train)
model_g.score(x_test,y_test)
```

Kernel

```python
model_linear_kernel=SVC(kernel='linear')
model_linear_kernel.fit(x_train,y_train)
model_linear_kernel.score(x_test,y_test)
```

random_forest

```python
from sklearn.ensemble import RandomForestClassifier
model=RandomForestClassifier()
model.fit(x_train,y_train)
```

KMeans

```python
from sklearn.cluster import KMeans
scaler = StandardScaler()
X_scaled = scaler.fit_transform(x)
```

```python
model = KMeans(n_clusters=4, random_state=42)
model.fit(X_scaled)

df['cluster'] = model.labels_
df
```

```python
pd.crosstab(df['efficiency_band'], df['cluster'])
```

```python
for k in range(2, 7):
    model = KMeans(n_clusters=k)
    model.fit(X_scaled)
    print(f"K={k}, Inertia={model.inertia_}")
```
